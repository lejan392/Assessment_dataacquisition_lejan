Q16

```python

df = pd.read_csv("student_performance.csv")

print("Missing values by column:\n", df.isna().sum())
numeric_columns = df.select_dtypes(include="number").columns
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].median())


score_columns = ["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]
df["Average_Score"] = df[score_columns].mean(axis=1)
df["Pass_Status"] = np.where(df["Average_Score"] >= 50, "Pass", "Fail")

print("Top 5 students by Average Score:\n",
      df.nlargest(5, "Average_Score"))
department_average = df.groupby("Department")["Average_Score"].mean()
print("Department with highest average score:",
      department_average.idxmax())
print(department_average.sort_values(ascending=False))
print("Average attendance by department:\n",
      df.groupby("Department")["Attendance"].mean())
print("Pass/fail counts:\n",
      df["Pass_Status"].value_counts())

summary = (
    df.groupby("Department")
      .agg(
          Total_Students=("Student_ID", "count"),
          Average_Score=("Average_Score", "mean"),
          Average_Attendance=("Attendance", "mean"),
          Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
          Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
      )
      .reset_index()
)

print("Department summary:\n", summary)
```

Q17

```python
sales_df=pd.read_csv('sales.csv')
print(sales_df)

sales_df["Order_Date"] = pd.to_datetime(sales_df["Order_Date"], errors="coerce")
print("Missing values:\n", sales_df.isna().sum())

sales_df = sales_df.dropna(subset=["Order_Date", "Product", "Category", "Region", "Sales", "Profit"])

sales_df["Sales"] = pd.to_numeric(sales_df["Sales"].astype(str).str.replace(',', ''), errors='coerce')
sales_df["Profit"] = pd.to_numeric(sales_df["Profit"].astype(str).str.replace(',', ''), errors='coerce')
sales_df["Discount"] = pd.to_numeric(sales_df["Discount"].fillna(0).astype(str).str.replace('%', ''), errors='coerce')

print("Total sales:", sales_df["Sales"].sum())
print("Total profit:", sales_df["Profit"].sum())
print("Average sales:", sales_df["Sales"].mean())

product_sales = (sales_df.groupby("Product")["Sales"].sum().sort_values(ascending=False))

print("Top 5 products by sales:\n", product_sales.head(5))

region_sales = sales_df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print("Sales by region:\n", region_sales)
print("Region with highest sales:", region_sales.idxmax())

category_summary = (
    sales_df.groupby("Category").agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Average_Discount=("Discount", "mean")
      ).sort_values("Total_Sales", ascending=False))

print("Category summary:\n", category_summary)

profit_by_category = sales_df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
print("Most profitable category:", profit_by_category.idxmax())

print("Product with highest sales:", product_sales.idxmax())
```
